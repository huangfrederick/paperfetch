## Search Batch of Papers by Keyword

```python
from paperfetch.openalex import search_works

papers = search_works("open access citation advantage", limit=10)
for paper in papers:
    status = "PDF" if paper.pdf_url else "no PDF"
    print(f"{paper.year}  [{status}]  {paper.title}")
```

## Batch download papers as PDFs

This will first check if the file already exists on disk first. If not, then publisher or repository links via `pdf_url` will be tried. Both of these attempts will be free since it doesn't query anything OpenAlex directly hosts.

If all fails, then OpenAlex's cached copies will be downloaded and incur API costs.

This will also check if the downloaded file is really a PDF instead of a plain HTML page.

```python
from paperfetch.download import download_pdf

papers = search_works("open access citation advantage", limit=5)
for paper in papers:
    path = download_pdf(paper, "../data/pdfs")
    print(paper.work_id, "->", path)
```
