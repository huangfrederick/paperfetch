## Load Keys

```python
import os

import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.environ["OPENALEX_API_KEY"]
print("Key loaded, length:", len(API_KEY))
```

## Search 1 paper by DOI

```python
response = requests.get(
    "https://api.openalex.org/works/doi:10.7717/peerj.4375",
    headers={"Authorization": f"Bearer {API_KEY}"},
    timeout=30,
)
response.raise_for_status()     # turns http error into python error.
work = response.json()

print(work["title"])
print(work["open_access"])
print(work["best_oa_location"])
```
