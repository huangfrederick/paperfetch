## Imports

```python
from pathlib import Path

import pandas as pd
import bibtexparser
from PyPaperBot.__main__ import start
```

```python
from bibtexparser.bparser import BibTexParser
from PyPaperBot.Paper import Paper


def safe_set_bibtex(self, bibtex):
    """Replacement for Paper.setBibtex that doesn't crash on unusual BibTeX."""
    self.bibtex = bibtex
    try:
        parser = BibTexParser(interpolate_strings=False)    # leave macros like july in place
        entry = bibtexparser.loads(bibtex, parser=parser).entries[0]
    except Exception:
        return  # keep the raw BibTeX, just skip extracting details

    self.year = entry.get("year", self.year)    # use year from bibtex if there is one
    self.authors = entry.get("author", self.authors)
    journal = entry.get("journal") or entry.get("publisher")
    if journal:
        self.jurnal = journal.replace("\\", "")


Paper.setBibtex = safe_set_bibtex
```

## Settings

```python
QUERY = "machine learning"
N_RESULTS = 3
OUT_DIR = Path("../data/first_search") # results location
OUT_DIR.mkdir(parents=True, exist_ok=True)
```

## Run Search

```python
start(
    query=QUERY,
    scholar_results=N_RESULTS,
    scholar_pages=range(1, 2),
    dwn_dir=str(OUT_DIR) + "/",
    proxy=[],
    restrict=0,     # 0: bibtex and metadata only, None: 0 and pdfs, 1:, pdfs only.
)
```
